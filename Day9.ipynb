```python
names = ["Amit", "Rahul", "Neha", "Karan"]
new_name = input("Enter the name to be searched : ")
for i in range(len(names)):
    if names[i] == new_name:
        print(new_name, "is present in the list at index " + str(names.index(new_name)))
        break
else:
    print(new_name, "is not present in the list")
```

```python
names = ["Amit", "Rahul", "Neha", "Karan"]
marks = [45, 82, 67, 91]
result = [(name, mark) for name, mark in zip(names,marks) if mark >= 50]
print("Passed:", result)
```

```python
names = ["Amit", "Rahul", "Neha", "Karan"]
marks = [45, 82, 67, 91]
for names, marks in zip(names,marks):
    print(names, "has scored", marks, "marks")
    if marks >=50:
        print(names, "has passed")
    else:
        print(names, "has failed")
```

```python
numbers = [-3, 2, -8, 4, 5, -1, 6]
new_list = [num ** 2 for num in numbers if num > 0]
print(new_list)
```

```python
numbers = [-3, 2, -8, 4, 5, -1, 6]
new_list = []
sqaure = 0
for num in numbers:
    if num > 0:
        sqaure = num ** 2
        new_list.append(sqaure)
print(new_list)
```

```python
numbers = [12, 5, 8, 19, 20, 31, 44]
even_num = [n for n in numbers if n % 2 == 0]
print(even_num)
```

```python
numbers = [12, 5, 8, 19, 20, 31, 44]
even_num = []
for num in numbers:
    if num%2 == 0:
        print(num, "is even")
        even_num.append(num)
print(even_num)
```
