```python
# find the longest sublist using a for loop

mylist = [[10, 20, 30, 40],
          [5, 15], 
          [100, 200, 50], 
          [7, 8, 9, 10, 11], 
          [25, 35, 45], 
          [1, 2, 3, 4, 5, 6]]
longestsublist = mylist[0]
for sublist in mylist :
    if len(sublist) > len(longestsublist):
        longestsublist = sublist
print("The longest sublist is : " , longestsublist)
    
```

```python
# find the sublist with the smallest minimum element

mylist = [[10, 20, 30, 40],
          [5, 15], 
          [100, 200, 50], 
          [7, 8, 9, 10, 11], 
          [25, 35, 45], 
          [1, 2, 3, 4, 5, 6]]
smallestmin = []
for sublist in mylist:
    minvalue = 10000
    for value in sublist:
        if value < minvalue:
            minvalue = value
    smallestmin.append(minvalue)
print("The list of minimum values of sublists is : " , smallestmin)
for value in smallestmin:
    if value < minvalue:
        minvalue = value
        position = smallestmin.index(minvalue)
print(mylist[position])
```
