```python
# nested list function
def find_largest_sublist(numbers):
    sumofsublist = []
    for row in numbers:
        total = 0
        for num in row:
            total += num
        sumofsublist.append(total)
    return sumofsublist

def find_largest_sum_in_sumofsublist(sumofsublist):
    lnum = 0
    for num in sumofsublist:
        if num > lnum:
            lnum = num
    return lnum

def find_sublist_as_per_largestsum(lnum, sumofsublist, numbers):
    position = sumofsublist.index(lnum)
    value = numbers[position]
    return value


numbers = [
    [10, 20, 30],
    [100, 50],
    [40, 60, 70, 80]
]
sumofsublist = find_largest_sublist(numbers)
lnum = find_largest_sum_in_sumofsublist(sumofsublist)
value = find_sublist_as_per_largestsum(lnum, sumofsublist, numbers)
print(f"The lis of sum of sublist is {sumofsublist}")
print(f"The largest sum in the sum of sublist is {lnum}")
print(f"The subset with largest sum is {value}")
```

```python
# Count numbers
def count_numbers(numbers):
    totalnumberscount = 0
    positivenumbers = 0
    negativenumbers = 0
    numberofzeroes = 0
    for num in numbers:
        totalnumberscount = totalnumberscount + 1
        if num > 0 :
            positivenumbers = positivenumbers + 1
        elif num == 0 :
            numberofzeroes = numberofzeroes + 1
        else :
            negativenumbers = negativenumbers + 1
    return totalnumberscount, positivenumbers, numberofzeroes, negativenumbers

numbers = [10, -5, 0, 20, -8, 0, 7, 0, -1, 9]
result = count_numbers(numbers)
totalnumberscount, positivenumbers, numberofzeroes, negativenumbers = result
print(f"Number of positive numbers is : {positivenumbers}")
print(f"Number of negative numbers is : {negativenumbers}")
print(f"Number of zeroes is : {numberofzeroes}")
print(f"Total number in list is : {totalnumberscount}")
```

```python
def list_statistics(numbers):
    # Finding Total
    total = 0
    for num in numbers:
        total = total + num
    # Finding average
    average = total / len(numbers)
    # Finding largest num
    lnum = 0
    for num in numbers :
        if num > lnum:
            lnum = num
    # Finding smallest number
    snum = 10000
    for num in numbers :
        if num < snum:
            snum = num
    return total, average, lnum, snum

numbers = [10,20,30,40]
result = list_statistics(numbers)
print(result)
total, average, lnum, snum = result
# it can also be defined as 
# total, average, lnum, snum = list_statistics(numbers)
print(f"The total of {numbers} is {total}")
print(f"The average of {numbers} is {average}")
print(f"The largest number in the list {numbers} is {lnum}")
print(f"The smallest number in the list {numbers} is {snum}")
```
