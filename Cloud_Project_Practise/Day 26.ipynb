```python
def decorator(function):
    def wrapper(role):
        print("Welcome to Program")
        if role == "admin":
            function()
        print("See you later")
        
    return wrapper

@decorator
def delete_vm():
    print("VM deleted")
role = input("Enter the role : ")
delete_vm(role)
```

```python
def decorator(function):
    count = 0
    def wrapper():
        nonlocal count
        count += 1
        print("Welcome to Program")
        print(f"Function called {count} time")
        function()
        print("See you later")
        
    return wrapper
@decorator
def program():
    print("Program is running")
program()
program()
```

```python
def log_function(function):
    def wrapper():
        print("Program started and the function going to be executed is : " , function.__name__ )
        function()
        print("Program ended")
    return wrapper

@log_function
def create_vm():
    print("VM created")
@log_function
def delete_vm():
    print("VM deleted")

create_vm()
delete_vm()
```

```python
def check_positive(function):
    def wrapper(number):
        print("Program started")
        function(number)
        print("program finished")
    return wrapper

@check_positive
def program(number):
    if number > 0:
        print("Program is executing and the number is : " , number)
    else:
        print("Number should be positive, retry please")
program(95)
```

```python
def calculate_log(function):
    def wrapper(*args):
        print("Calculation Started")
        function(*args)
        print("Calculation completed")
    return wrapper

@calculate_log
def add_numbers(*args):
    sum = 0
    for num in args:
        sum = sum + num
    print("Sum is : " , sum)
add_numbers(10,20)
```
