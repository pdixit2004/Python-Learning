```python
transactions = [
    ["Rahul", 5000],
    ["Priya", -2000],
    ["Amit", 7000],
    ["Neha", -1000],
    ["Ravi", 3000]
]

def trans_filter(transactions):

    for row in transactions:
        if row[1] > 0:
            yield row

for value in trans_filter(transactions):
    print(value)
```

```python
resources = [
    ["VM1", 100],
    ["Storage1", 50],
    ["VM2", 200],
    ["Database1", 150]
]

def conditional_resources(resources):

    for row in resources:
        if row[1] > 100:
            yield row[0]

for value in conditional_resources(resources):
    print(value)

```

```python
def generate_vm_names():

    for name in range(1,6):
        yield name

for numbers in generate_vm_names():
    print(f"VM - 0{numbers}")
```

```python
employees = [
    ["Rahul", 50000],
    ["Priya", 75000],
    ["Amit", 45000],
    ["Neha", 90000]
]

def conditional_employee(employees):

    for row in employees:
        if row[1] > 60000:
            yield row[0]

for value in conditional_employee(employees):
    print(value)
```

```python
numbers = [10, 70, 30, 90, 60, 20]

def conditional_number(numbers):

    for num in numbers:
        if  num > 50:
            yield num

for value in conditional_number(numbers):
    print(value)
```

```python
def even_numbers():

    for num in range(1,21):
        if num % 2 == 0:
            yield num

for number in even_numbers():
    print(number)
```

```python
def count_numbers():

    for num in range(1,6):
        yield num
        
for number in count_numbers():
    print(number)
```
