```python
from abc import ABC, abstractmethod

class CloudResource(ABC):

    def __init__(self, resource_id, name, region):
        self.resource_id = resource_id
        self.name = name 
        self.region = region

    def display(self):
        print("Resource id is : ", self.resource_id)
        print("Name is : " , self.name)
        print("Region is : " , self.region)

    @abstractmethod
    def calculate_cost(self):
        pass

class Virtualmachine(CloudResource):
    def __init__(self, resource_id, name, region, hours, hourly_rate):
        super().__init__(resource_id, name,region)
        self.hours = hours
        self.hourly_rate = hourly_rate

    def calculate_cost(self):
        costvm = self.hours * self.hourly_rate
        return costvm
    
class Storage(CloudResource):
    def __init__(self, resource_id, name, region, cost_per_gb, hourly_cost):
        super().__init__(resource_id, name,region)
        self.cost_per_gb = cost_per_gb
        self.hourly_cost = hourly_cost

    def calculate_cost(self):
        costs = self.cost_per_gb * self.hourly_cost
        return costs

class Database(CloudResource):
    def __init__(self, resource_id, name, region, cost_db_per_hour, hourly):
        super().__init__(resource_id, name,region)
        self.cost_db_per_hour = cost_db_per_hour
        self.hourly = hourly

    def calculate_cost(self):
        costdb = self.cost_db_per_hour * self.hourly
        return costdb
    
class CloudManager:
    def __init__(self):
        self.resource_list = []
    #Add a resource
    def add(self,resources):
        return self.resource_list.append(resources)
    #Remove a resource
    def remove(self,resources):
        return self.resource_list.remove(resources)
    # find a resource using resource_id
    def find(self , resource_id):
        for resource in self.resource_list:
            if resource.resource_id == resource_id:
                return resource
        return None
    # calculate total cost
    def total_cost(self):
        total = 0
        for resource in self.resource_list:
            total = total + resource.calculate_cost()
        return total
    #find highest cost resource
    def highest_cost(self):
        highest = None
        highest_cost_resource = 0
        for resource in self.resource_list:
            if highest is None or highest < resource.calculate_cost():
                highest = resource.calculate_cost()
                highest_cost_resource = resource
        return highest, highest_cost_resource
    #find lowest cost resource
    def lowest_cost(self):
        lowest = None
        lowest_cost_resource = 0
        for resource in self.resource_list:
            if lowest is None or lowest > resource.calculate_cost():
                lowest = resource.calculate_cost()
                lowest_cost_resource = resource
        return lowest, lowest_cost_resource
    #filter by region
    def filter_region(self,region):
        result = []
        for resource in self.resource_list:
            if resource.region == region:
                result.append(resource)
        return result
    def __len__(self):
        return len(self.resource_list)

    
resource = [Virtualmachine("VM123", "VM01" , "East US" , 120, 100), Storage("Stor123", "Storage01", "West US", 150, 100), Database("DbS123" , "DB01", "Central India", 200, 120)]
resource_list = []
for value in resource:
    resource_list.append(value)
    value.display()
    print(value.calculate_cost())
manager = CloudManager()
for value in resource_list:
    manager.add(value)
print(f"The total of all resources is : {manager.total_cost()}")
highest_cost, highest_resource = manager.highest_cost()
print(f"The highest cost of resource is : {highest_cost}")
print(f"It belongs to resource : {highest_resource.name}")
lowest_cost, lowest_resource = manager.lowest_cost()
print(f"The highest cost of resource is : {lowest_cost}")
print(f"It belongs to resource : {lowest_resource.name}")
print(len(manager))
```

```python
from abc import ABC, abstractmethod

class Payment(ABC):

    @abstractmethod
    def pay(self):
        pass

class CreditCardPayment(Payment):

    def pay(self):
        return "This is a credit card payment. Thank you"

class UPIPayment(Payment):

    def pay(self):
        return "This is a UPI payment. Thank you"

class BankTransferPayment(Payment):

    def pay(self):
        return "This is a bank transfer payment. Thank you"


payments = [CreditCardPayment(), UPIPayment(), BankTransferPayment()]

for value in payments:
    print(value.pay())
```

```python
from abc import ABC, abstractmethod

class Employee(ABC):

    @abstractmethod
    def calculate_salary(self):
        pass

class FullTimeEmployee(Employee):

    def __init__(self, monthly_salary):
        self.monthly_salary = monthly_salary
    def calculate_salary(self):
        return self.monthly_salary

class PartTimeEmployee(Employee):

    def __init__(self, hours, hourly):
        self.hours = hours
        self.hourly = hourly
    def calculate_salary(self):
        salary = self.hours * self.hourly
        return salary

f1 = FullTimeEmployee(30000)
p1 = PartTimeEmployee(20, 1000)
print(f1.calculate_salary())
print(p1.calculate_salary())
```
