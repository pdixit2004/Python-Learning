```python
class User:
    def __init__(self,name,user_ID):
        self.name = name 
        self.user_ID = user_ID
    def send_message(self):
        print("Hi everyone")
    def join_chatroom(self):
        print(f"{self.name} joined the chatroom")
    def leave_chatroom(self):
        print(f"{self.name} left the chatroom")

class Message:
    def __init__(self,sender,message, time):
        self.sender = sender
        self.message = message
        self.time = time
    def display(self):
        print(f"{self.sender} has send {self.message} at {self.time}")

class ChatRoom:
    def __init__(self, chat_room_name, user_list, message_list):
        self.chat_room_name = chat_room_name
        self.user_list = user_list
        self.message_list = message_list
    def add_user(self, user):
        if user not in self.user_list:
            self.user_list.append(user)
        return self.user_list
    def remove_user(self, user):
        if user in self.user_list:
            self.user_list.remove(user)
        return self.user_list
    def add_message(self,message):
        self.message = message
        self.message_list.append(message)
        return self.message_list
    def display_message(self):
        print(f"The messages are : {self.message_list}")

u1 = User("prakhar", 1)
u2 = User("Aniket", 2)
u3 = User("ABC", 3)
u1.join_chatroom()
u1.send_message()
u2.join_chatroom()
u2.send_message()
u3.join_chatroom()
u3.send_message()
```

```python
class Herbivore:
    def __init__(self, plant_food, **kwargs):
        self.plant_food = plant_food
        super().__init__(**kwargs)

    def eat_plants(self):
        print(f"This animal eats plants such as {self.plant_food}")


class Carnivore:
    def __init__(self, meat_food, **kwargs):
        self.meat_food = meat_food
        super().__init__(**kwargs)

    def eat_meats(self):
        print(f"This animal eats meat such as {self.meat_food}")


class Omnivore:
    def __init__(self, food_type, **kwargs):
        self.food_type = food_type
        super().__init__(**kwargs)

    def eat_both(self):
        print(f"This animal eats both plants and meat: {self.food_type}")


class Bear(Herbivore, Carnivore, Omnivore):
    def __init__(self, size, plant_food, meat_food, food_type):
        self.size = size
        super().__init__(
            plant_food=plant_food,
            meat_food=meat_food,
            food_type=food_type
        )

    def display(self):
        print(f"The size of bear is {self.size}")
        print(f"The plant that bear eats is: {self.plant_food}")
        print(f"The meat that bear eats is: {self.meat_food}")
        print(f"The food type is: {self.food_type}")


b1 = Bear("500 kg", "Berries", "Fish", "Plants and seafood")

b1.display()
b1.eat_plants()
b1.eat_meats()
b1.eat_both()
```

```python
class Herbivore:
    def __init__(self, plant_food):
        self.plant_food = plant_food
    def eat_plants(self):
        print(f"this animal eat plants such as {self.plant_food}")

class Carnivore:
    def __init__(self, meat_food):
        self.meat_food = meat_food
    def eat_meats(self):
        print(f"this animal eat meats such as {self.meat_food}")

class Omnivore:
    def __init__(self, food_type):
        self.food_type = food_type
    def eat_both(self):
        print(f"this animal eat both plants & meats such as {self.food_type}")

class Bear(Herbivore, Carnivore, Omnivore):
    def __init__(self, size, plant_food, meat_food, food_type):
        self.size = size
        Herbivore.__init__(self,plant_food)
        Carnivore.__init__(self,meat_food)
        Omnivore.__init__(self,food_type)
    def display(self):
        print(f"The size of bear is {self.size}")
        print(f"The plant that bear eats is {self.plant_food}")
        print(f"The meat that bear eats is : {self.meat_food}")
        print(f"The both types of food that bear eats is : {self.food_type}")

b1 = Bear("500 kg","Berries", "Fish", "plants & sea food")
b1.display()
```

```python
from abc import ABC, abstractmethod

class Employee(ABC):

    @abstractmethod
    def calculate_salary(self):
        pass

class Intern(Employee):
    def __init__(self, salary):
        self.salary = salary
    def calculate_salary(self):
        print(f"He gets a fixed stipend : {self.salary}")

class FullTimeEmployee(Employee):

    def __init__(self, basic, allowance):
        self.basic = basic
        self.allowance = allowance

    def calculate_salary(self):
        self.salary = self.basic + self.allowance
        return self.salary

class ContractEmployee(Employee):

    def __init__(self, hours, wages_per_hour):
        self.hours = hours
        self.wages_per_hour = wages_per_hour
    def calculate_salary(self):
        self.salary = self.hours * self.wages_per_hour
        return self.salary

E1 = FullTimeEmployee(18000, 5000)
E1.calculate_salary()
print(f"The salary of full time employee is : {E1.calculate_salary()}")
```

```python
class Vehicle:
    def __init__(self, brand,model):
        self.brand = brand
        self.model = model
    def display(self):
        print(f"The brand is : {self.brand}")
        print(f"The model is : {self.model}")

class Car(Vehicle):
    def __init__(self, brand,model,seats):
        super().__init__(brand,model)
        self.seats = seats
    def display(self):
        super().display()
        print(f"The seats in car are : {self.seats}")

car1 = Car("Maruti", "800", 4)
car1.display()
```

```python
class Student:
    def __init__(self, name,roll_num, marks):
        self._name = name
        self._roll_num = roll_num
        self._marks = marks

    @property
    def name(self):
        return self._name
    @name.setter
    def name(self, value):
        if value == "":
            print("Name does not exist")
        else:
            self._name = value

    @property
    def roll_num(self):
        return self._roll_num
    @roll_num.setter
    def roll_num(self, value):
        if value >= 1 and value <= 100:
            self._roll_num = value
        else:
            print("Roll number does not exist")

    @property
    def marks(self):
        return self._marks
    @marks.setter
    def marks(self, value):
        if value < 0:
            print("Marks cannot be negative")
        else:
            self._marks = value

s1 = Student("Prakhar", 20, 99)
s2 = Student("Aniket", 35, 98)
print(s1.name)
print(s1.roll_num)
s1.name = "ABC"
print(s1.name)
```

```python
class Book:
    def __init__(self,title,author):
        self.title = title
        self.author = author
        self.review_list = []
    def display(self):
        print(f"The title of book is : {self.title}")
        print(f"The author of book is : {self.author}")
        print(f"The reviews list is : {self.review_list}")
    def add_review(self, review):
        self.review_list.append(review)
        return self.review_list
    def count_reviews(self):
        count = 0
        for review in self.review_list:
            count += 1
        return count

book1 = Book("The End", "Prakhar")
book1.add_review("This book is perfect")
book1.display()
book1.count_reviews()
book2 = Book("The Start", "ABC")
book2.add_review("This book is awesome")
book2.display()
book2.count_reviews()

```

```python
class BankAccount:
    def __init__(self, acc_num, acc_holder, balance):
        self.acc_num = acc_num
        self.acc_holder = acc_holder
        self.balance = balance
    def deposit_money(self, money):
        self.money = money
        self.balance = self.money + self.balance
        return self.balance
    def withdraw_money(self, w_money):
        self.w_money = w_money
        if self.balance >= self.w_money:
            self.balance = self.balance - self.w_money
        else:
            print(f"You cannot withdraw amount more than balance, balance is : {self.balance}")
        return self.balance
    def display(self):
        print(f"The account number is : {self.acc_num}")
        print(f"The account holder name is : {self.acc_holder}")
        print(f"The balance is : {self.balance}")

acc1 = BankAccount(101, "Prakhar", 1_00_000)
acc1.display()
acc1.withdraw_money(1000000000)


```
