```python
enter_string = input("Enter the string : ")
set1 = set()
count = 0
for ch in enter_string:
    if ch not in set1:
        set1.add(ch)
        count += 1
print(f"The number of unique characters in this string are {set1} and the count is {count}")
```

```python
list = [10, 20, 30, 20, 40, 50, 40, 30, 60, 80, 60, 70]
set1 = set()
set2 = set()
for num in list:
    if num not in set1:
        set1.add(num)
    else:
        set2.add(num)
print(f"The duplicate elements are : {set2}")
```

```python
list1 = [1, 2, 3, 4, 5] 
set1 = set(list1)
list2 = [5, 6, 7, 8]
set2 = set(list2)
result = set1.intersection(set2)
if result == set():
    print("The sets have no common element")
else:
    print(f"The common elements are : {result}")
```

```python
enter_string = input("Enter a string : ")
counter = 0
for ch in enter_string:
    if ch == " ":
        counter += 1
print(f"The spaces in the string is : {counter}")
```

```python
#cleaner version

row = 5

for i in range(row):

    for j in range(row):

        if i == 0 or i == row - 1 or j == 0 or j == row - 1:
            print("*", end="")
        else:
            print(" ", end="")

    print()
```
