```python
def highest_total_marks(students):
    highesttotal = 0
    for row in students:
        total = 0
        for marks in row[1:]:
            total = total + marks
            if total > highesttotal:
                highesttotal = total
        print(f"The total marks of student {row[0]} is {total}")
    print(f"the highest total is : {highesttotal}")
    


students = [
    ["Rahul", 80, 90],
    ["Priya", 95, 85],
    ["Amit", 70, 75, 70],
    ["Neha", 90, 98]
]

highest_total_marks(students)
```

```python
def calculate_total(employees):
    totallist = []
    for row in employees :
        total = 0
        for marks in row[1:]:
            total = total + marks
        totallist.append(total)
    return totallist

def calculate_average(totallist, employees):
    averagelist = []
    for y in range(len(totallist)):
        average = round(totallist[y] / (len(employees[y]) - 1) , 1)
        averagelist.append(average)
    return averagelist

def get_grade(averagelist):
    Gradelist = []
    for value in averagelist:
        if value > 90:
            Grade = "A+"
        elif value >= 80 or value == 90:
            Grade = "B"
        elif value >= 70:
            Grade = "C"
        elif value >= 60:
            Grade = "D"
        else:
            Grade = "F"
        Gradelist.append(Grade)
    return Gradelist

def highest_lowest_employee(employees,averagelist):
    highestaverage = None
    positionhavg = 0
    positionlavg = 0
    lowestaverage = None
    for value in averagelist:
        if highestaverage is None or value > highestaverage :
            highestaverage = value
            positionhavg = averagelist.index(highestaverage)
        if lowestaverage is None or value < lowestaverage :
            lowestaverage = value
            positionlavg = averagelist.index(lowestaverage)
    print(f"The highest average is : {highestaverage}")
    print(f"The employee with highestaverage is : {employees[positionhavg]}")
    print(f"The lowest average is : {lowestaverage}")
    print(f"The employee with lowestaverage is : {employees[positionlavg]}")
    print()

def analyze_employees(employees, totallist,averagelist,Gradelist):
    for i in range(len(employees)):
        print(f"Employee name is : {employees[i][0]}")
        print(f"The total marks of subjects is : {totallist[i]}")
        print(f"The average of this employee is : {averagelist[i]}")
        print(f"The Grade of this employee is : {Gradelist[i]}")
        print()


employees = [
    ["Rahul", 80, 75, 90],
    ["Priya", 95, 88, 92],
    ["Amit", 60, 70, 65],
    ["Neha", 90, 95, 98],
    ["Ravi", 72, 68, 75]
]

totallist = calculate_total(employees)
averagelist = calculate_average(totallist, employees)
Gradelist = get_grade(averagelist)
analyze_employees(employees, totallist,averagelist,Gradelist)
highest_lowest_employee(employees,averagelist)
```

```python
def get_high_salary_employees(employees, threshold):
    get_high_salary = []
    for row in employees:
        if row[1] > threshold:
            get_high_salary.append(row)
    return get_high_salary


employees = [
    ["Rahul", 50000],
    ["Priya", 75000],
    ["Amit", 45000],
    ["Neha", 90000],
    ["Ravi", 65000]
]
threshold = 70000
result = get_high_salary_employees(employees, 70000)
print(f"The employees whose salary is greater than {threshold} are : {result}")

```

```python
def analyze_numbers(numbers):
    total = 0
    average = 0
    largest = None
    smallest = None
    counteven = 0
    countodd = 0
    for num in numbers:
        total = total + num
        if largest is None or num > largest:
            largest = num
        if smallest is None or num < smallest:
            smallest = num
        if num % 2 == 0:
            counteven = counteven + 1
        if num % 2 != 0:
            countodd = countodd + 1
    average = round(total / len(numbers) , 2)
    print("Average is : " , average)
    print("Total is : " , total)
    print("Smallest num is : " , smallest)
    print("Largest num is : " , largest)
    print("Number of even numbers are : " , counteven)
    print("Number of odd numbers are : " , countodd)


numbers = [10, 25, -5, 40, 33, 12, 10]
analyze_numbers(numbers)
```
