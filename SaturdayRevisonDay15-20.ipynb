```python
def process_transactions(transactions):
    total = 0
    count = 0
    for value in transactions:
        try:
            num = value[1]
            if type(num) == str :
                raise ValueError("transaction is a string")
            if num <= 0:
                raise ValueError("transaction is negative & invalid, enter again")
            count = count + 1
            total = total + num
            print("Valid transaction " , num)
        except ValueError as error:
            print("Invalid transaction" , value)
            print("Reason : " , error)
    return count, total
        

transactions = [
    ["Rahul", 5000],
    ["Priya", 3000],
    ["Amit", -1000],
    ["Neha", 7000],
    ["Ravi", "5000"],
    ["Karan", 0]
]
count, total = process_transactions(transactions)

print("Count is : " , count)
print("Total is : ", total)

```

```python
# employee analyzer
def calculate_gross(employees):
    gross_list = []
    for row in employees:
        grosssalary = 0
        for value in row[1:]:
            grosssalary = grosssalary + value
        gross_list.append(grosssalary)
    return gross_list

def find_highest_employee(gross_list,employees):
    highest_gross_salary = None
    for salary in gross_list:
        if highest_gross_salary is None or salary > highest_gross_salary:
            highest_gross_salary = salary
            position = gross_list.index(highest_salary)
    print(f"The employee with highest salary is : {employees[position]} and the highest salary is : {highest_gross_salary}")
    

def find_lowest_employee(gross_list, employees):
    lowest_gross_salary = None
    for salary in gross_list:
        if lowest_gross_salary is None or salary < lowest_gross_salary:
            lowest_gross_salary = salary
            position = gross_list.index(lowest_salary)
    print(f"The employee with lowest salary is : {employees[position]} and the lowest salary is : {lowest_gross_salary}")


def calculate_average_gross(gross_list, employees):
    total_gross = 0
    average_salary = 0
    for salary in gross_list:
        total_gross = total_gross + salary
    average_salary = total_gross / len(employees)
    return average_salary

def get_high_earners(gross_list):
    position = 0
    for salary in gross_list:
        if salary >= 70000:
            position = gross_list.index(salary)
            print(f"The high earner employees are : {employees[position]}")

employees = [
    ["Rahul", 50000, 5000],
    ["Priya", 70000, 10000],
    ["Amit", 45000, 3000],
    ["Neha", 80000, 5000],
    ["Ravi", 60000, 7000]
]
gross_list = calculate_gross(employees)
average = calculate_average_gross(gross_list, employees)

print(f"The gross list salary of employees is : {gross_list}")
print()
find_highest_employee(gross_list,employees)
print()
find_lowest_employee(gross_list,employees)
print()
print(f"The average salary of all employees is : {average}")
print()
get_high_earners(gross_list)
```

```python
def analyze_numbers(numbers):
    countp = 0
    countn = 0
    countz = 0
    for num in numbers:
        if num == 0 :
            print(f"{num} is a zero")
            countz = countz + 1
        elif num > 0 :
            print(f"{num} is positive")
            countp = countp + 1
        else:
            print(f"{num} is negative")
            countn = countn + 1
    return countz, countn, countp

numbers = [10, -5, 0, 20, -2, 0]
countz, countp, countn = analyze_numbers(numbers)
print(f"Number of positive numbers are : {countp}")
print(f"Number of negative numbers are : {countn}")
print(f"Number of zeroes in numbers list are : {countz}")
```

```python
def find_largest(numbers):
    lnum = None
    for num in numbers:
        if lnum is None or num > lnum :
            lnum = num
    print(f"Largest number is {lnum}")

numbers = [10, -2, 0, 100, 48, -90, 0, -43, 576]
find_largest(numbers)
```
