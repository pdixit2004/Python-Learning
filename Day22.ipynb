```python
class CloudResource:
    def __init__(self,name, region, resource_id):
        self.name = name
        self.region = region
        self.resource_id = resource_id
    def display(self):
        print(f"Name is : {self.name}")
        print(f"Region is : {self.region}")
        print(f"Resource id is : {self.resource_id}")

class VM(CloudResource):
    def __init__(self,name, region, resource_id,sku,hours,hourly_cost):
        super().__init__(name, region, resource_id)
        self.sku = sku
        self.hours = hours
        self.hourly_cost = hourly_cost
    def cost(self):
        cost_VM = 0
        cost_VM = self.hourly_cost * self.hours
        return cost_VM
    def display(self):
        super().display()
        print(f"VM SKU     : {self.sku}")
class Storage(CloudResource):
    def __init__(self,name, region, resource_id,storage_type, size_gb, price_per_gb):
        super().__init__(name,region,resource_id)
        self.storage_type = storage_type
        self.size_gb = size_gb
        self.price_per_gb = price_per_gb
    def cost(self):
        cost_Storage = 0
        cost_Storage = self.size_gb * self.price_per_gb
        return cost_Storage
    def display(self):
        super().display()
        print(f"Storage resource       : {self.storage_type}")
class Database(CloudResource):
    def __init__(self,name, region, resource_id,engine,hours,hourly_cost):
        super().__init__(name,region,resource_id)
        self.engine = engine
        self.hours = hours
        self.hourly_cost = hourly_cost
    def cost(self):
        cost_DB = 0
        cost_DB = self.hours * self.hourly_cost
        return cost_DB
    def display(self):
        super().display()
        print(f"Databse      : {self.engine}")

class CloudManager:
    def __init__(self):
        self.resource_list = []
    # Add a resource
    def add_resources(self, resources):
        self.resource_list.append(resources)
    # Remove a resource
    def remove_resources(self, resources):
        self.resource_list.remove(resources)
    # Total cost of resources
    def total_cost(self):
        total = 0
        for resource in self.resource_list:
            total = total + self.resource.cost()
        return total
    # Find a resource
    def find_resources(self, resource_id):
        for resource in self.resource_list:
            if resource.resource_id == resource_id:
                return resource
            return None
    # Find highest cost of a resource
    def find_highest_cost(self, resources):
        highest_cost = None
        if highest_cost is None or highest_cost < resources.cost:
            highest_cost = resources.cost
        return highest_cost
    # Find lowest cost of a resource
    def find_lowest_cost(self, resources):
        lowest_cost = None
        if lowest_cost is None or lowest_cost > resources.cost:
            lowest_cost_cost = resources.cost
        return lowest_cost
    # Filter region 
    def filter_region(self, region):
        result = []

        for resource in self.resource_list:
            if resource.region == region:
                result.append(resource)

        return result
    #Filter by type
    def filter_type(self, resource_type):
        result = []

        for resource in self.resource_list:
            if isinstance(resource, resource_type):
                result.append(resource)

        return result


vm1 = VM("VM01", "Central India", "VM123", "Standard_D4s_v5", 20, 2)
storage1 = Storage("Storage01", "Central India", "Storage123", "Blob Storage", 1024, 1000)
db1 = Database("DB01", "East US", "DB123", "MySQL", 1000, 250)
```

```python
class BankAccount:
    def __init__(self, account_holder,balance):
        self.account_holder = account_holder
        self._balance = balance

    @property
    def balance(self):
        return self._balance
    @balance.setter
    def balance(self,value):
        if value < 0:
            raise ValueError("balance cannot be negative")
        self._balance = value

ba1 = BankAccount("Prakhar", 1000000)
print(ba1.balance)
ba1.balance = 750000
print(ba1.balance)
```

```python
class BankAccount:
    def __init__(self,balance):
        self.__balance = balance

    def deposit(self,amount):
        if amount > 0:
            self.__balance = self.__balance + amount
        else:
            raise ValueError

    def withdraw(self,withdraw_amount):
        if withdraw_amount > 0 and withdraw_amount <= self.__balance:
            self.__balance = self.__balance - withdraw_amount
        else:
            raise ValueError
    def get_balance(self):
        print(f"The final balance after deposit & withdrawl is : {self.__balance}")

user1 = BankAccount(1000)
user1.deposit(500)
user1.withdraw(200)
user1.get_balance()

```

```python
import math

class Shape:
    def area(self):
        print("This function defines the area")

class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius
    def area(self):
        return math.pi * self.radius * self.radius

class Rectangle(Shape):
    def __init__(self,length,width):
        self.length = length
        self.width = width
    def area(self):
        return self.length * self.width

c1 = Circle(5)
r1 = Rectangle(5,10)
print(f"Circle area is : {c1.area()}")
print(f"Rectangle area is : {r1.area()}")
```

```python
# using super()

class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary


class Manager(Employee):
    def __init__(self,name,salary,team_size):
        super().__init__(name,salary)
        self.team_size = team_size

manager = Manager("Prakhar", 850000, 15)
print(f"Employee name is : {manager.name}")
print(f"Employee salary is : {manager.salary}")
print(f"Employee team size is : {manager.team_size}")
```
