```python
def gross_salary(employees):
    grosslist = []
    for row in employees:
        total = 0
        for value in row[1:] :
            total = total + value      
        grosslist.append(total)
    return grosslist

def calculate_tax(grosslist):
    taxlist = []
    taxonsalary = 0
    for salary in grosslist:
        if salary < 50000:
            tax = 0
        elif salary < 70000:
            tax = salary * 0.10
        elif salary < 100000:
            tax = salary * 0.20
        else:
            tax = salary * 0.30
        taxlist.append(tax)
    return taxlist

def find_salary(grosslist, taxlist):
    salarylist = []
    for i in range(len(grosslist)):
        salary = grosslist[i] - taxlist[i]
        salarylist.append(salary)
    return salarylist

def analyze_employee_data(employees, grosslist, taxlist, salarylist):
    highestgross = 0
    avggrosssalary = 0
    smallestgross = grosslist[0]
    grosssum = 0
    for value in grosslist:
        grosssum = grosssum + value
        if value < smallestgross:
            smallestgross = value
            positionsg = grosslist.index(smallestgross)
        if value > highestgross:
            highestgross = value
        positionhg = grosslist.index(highestgross)
    avggrosssalary = grosssum / len(grosslist)
    print(f"The highest gross salary is : {highestgross}")
    print(f"The employee with highest gross salary is {employees[positionhg]}")
    print(f"The smallest gross salary is : {smallestgross}")
    print(f"The employee with smallest gross salary is {employees[positionsg]}")
    print(f"The avg gross salary is : {avggrosssalary}")

employees = [
    ["Rahul", 50000, 5000, 3000],
    ["Priya", 60000, 7000, 4000],
    ["Amit", 45000, 3000, 2000],
    ["Neha", 75000, 8000, 5000]
]
grosslist = gross_salary(employees)
print(grosslist)
taxlist = calculate_tax(grosslist)
print(taxlist)
salarylist = find_salary(grosslist, taxlist)
print(salarylist)
analyze_employee_data(employees, grosslist, taxlist, salarylist)

```

```python
def calculate_sales(data):
    totallist = []
    averagemonthlysaleslist = []
    lsaleslist = []
    ssaleslist = []
    for row in data:
        lsales = 0
        ssales = row[1]
        total = 0
        average = 0
        for value in row[1:] :
            total = total + value
            if value > lsales:
                lsales = value
            if value < ssales:
                ssales = value
        average = round(total / (len(row) - 1) , 2)        
        totallist.append(total)
        averagemonthlysaleslist.append(average)
        lsaleslist.append(lsales)
        ssaleslist.append(ssales)
    return totallist, averagemonthlysaleslist, lsaleslist, ssaleslist

def performance_level(averagemonthlysaleslist):
    performance_list = []
    for avg in range(len(averagemonthlysaleslist)):
        if averagemonthlysaleslist[avg] >= 1800 :
            Performance = "Excellent"
            performance_list.append(Performance)
        elif averagemonthlysaleslist[avg] >= 1400 :
            Performance = "Good"
            performance_list.append(Performance)
        elif averagemonthlysaleslist[avg] >= 1000 :
            Performance = "Average"
            performance_list.append(Performance)
        else :
            Performance = "Needs Improvement"
            performance_list.append(Performance)
    return performance_list

def find_top_employee(totallist, data):
    total = 0
    position = 0
    for i in range(len(totallist)):
        if totallist[i] > total:
            total = totallist[i]
    position = totallist.index(total)
    print(f"The employee with highest total sales is {data[position]}")



data = [
    ["Rahul", 1200, 1500, 900, 1800],
    ["Priya", 2000, 1700, 2100, 1900],
    ["Amit", 800, 950, 700, 1100],
    ["Neha", 1500, 1300, 1600, 1400]
]
results = calculate_sales(data)
totallist, averagemonthlysaleslist, lsaleslist, ssaleslist = results
performance_list = performance_level(averagemonthlysaleslist)
top_employee = find_top_employee(totallist, data)
print()

for i in range(len(data)):
    print(f"Employee: {data[i][0]}")
    print(f"Total sales: {totallist[i]}")
    print(f"Average monthly sales: {averagemonthlysaleslist[i]}")
    print(f"Highest monthly sale: {lsaleslist[i]}")
    print(f"Lowest monthly sale: {ssaleslist[i]}")
    print(f"Performance: {performance_list[i]}")
    print()

```

```python
def analyze_numbers(numbers):
    sump = 0
    sumn = 0
    sumzero = 0
    countp = 0
    countn = 0
    countzero = 0
    for num in numbers :
        if num > 0 :
            countp = countp + 1
            sump = sump + num
        elif num == 0 :
            countzero = countzero + 1
        else:
            countn = countn + 1
            sumn = sumn + num
    return sump , countp, sumn, countn, sumzero, countzero

numbers = [10, -5, 0, 20, -8, 0, 15, -3]
sump, countp, sumn, countn, sumzero, countzero = analyze_numbers(numbers)
print(f"The number of positive numbers is {countp}")
print(f"The number of negative numbers is {countn}")
print(f"The number of zero count is {countzero}")
print(f"The sum of positive numbers is {sump}")
print(f"The sum of negative numbers is {sumn}")
```
