```python
# mini Project - Student Marks

students = [
    [80, 75, 90],
    [60, 85, 70],
    [95, 90, 92]
]
student_number = 1
highestmark = 0
totalmarkslist = []
for student in students:
    total = 0
    for mark in student:
        total = total + mark
    print("Student" , student_number ,  "total marks : ", total)
    totalmarkslist.append(total)
    average = round(total / len(student) , 2)
    print("Student" , student_number ,  "average marks : ", average)
    student_number = student_number + 1
print("The list of total marks is : ", totalmarkslist)
for totalmarks in totalmarkslist:
    if totalmarks > highestmark:
        highestmark = totalmarks
print("The highest marks is : ", highestmark)
print("The student number with highest marks is : ", totalmarkslist.index(highestmark) + 1)
```

```python
# row sums
numbers = [
    [10, 20, 30],
    [5, 15, 25],
    [2, 4, 6]
]
for row in numbers:
    sum = 0
    for value in row:
        sum = sum + value
    print("Row of numbers:", row, "has sum:", sum)
```

```python
# find the largest number
numbers = [
    [12, 45, 7],
    [89, 23, 56],
    [34, 91, 18]
]
lnum = 0
for row in numbers:
    for value in row:
        if value > lnum:
            lnum = value
print("The largest number is:", lnum)
```
