```python
class Employee:
    def __init__(self,name,salary,score):
        self.name = name
        self.salary = salary
        self.score = score

    def get_rating(self):
        rating = 0
        if self.score >= 90:
            rating = "Excellent"
        elif self.score >= 75 :
            rating = "Good"
        elif self.score >= 60 :
            rating = "Average"
        else:
            rating = "Needs improvment"
        return rating

    def calculate_bonus(self):
        bonus = 0
        if self.get_rating() == "Excellent":
            bonus = 0.2 * self.salary
        elif self.get_rating() == "Good":
            bonus = 0.1 * self.salary
        elif self.get_rating() == "Average":
            bonus = 0.05 * self.salary
        else:
            bonus = 0
        return bonus

    def display(self):
        print(f"The employee name is : {self.name}")
        print(f"The employee salary is : {self.salary}")
        print(f"The employee score is : {self.score}")
        print(f"The rating of employee is : {self.get_rating()}")
        print(f"The bonus of employee is : {self.calculate_bonus()}")

count = 0
emp= []
bonus_list = []
emp_bonus = 0
emp1 = Employee("Prakhar", 85000, 92)
emp.append(emp1)
count = count+1
emp2 = Employee("Amit", 80000, 78)
emp.append(emp2)
count = count+1
emp3 = Employee("Rachna", 62000, 85)
emp.append(emp3)
count = count+1
emp4 = Employee("Aniket", 135000, 80)
emp.append(emp4)
count = count+1
emp5 = Employee("Neha", 75000, 62)
emp.append(emp5)
count = count+1
highest_performer = None
lowest_performer = None
highest_performer_value = 0
lowest_performer_value = 0
total_performance_score = 0
average_performance_score = 0
emp_bonus = 0
total_bonus = 0
for value in emp:
    emp_bonus = value.calculate_bonus()
    bonus_list.append(emp_bonus)
    total_performance_score = total_performance_score + value.score
    if highest_performer is None or value.score > highest_performer:
        highest_performer = value.score
        highest_performer_value = value.name
    if lowest_performer is None or value.score < lowest_performer:
        lowest_performer = value.score
        lowest_performer_value = value.name
    value.display()
    print()
average_performance_score = total_performance_score / count
for bonus in bonus_list:
    total_bonus = total_bonus + bonus

print(f"The total bonus of all employees is : {total_bonus}")
print(f"The highest performance score is : {highest_performer} and the employee who has achieved it is {highest_performer_value}")
print(f"The lowest performance score is : {lowest_performer} and the employee who has it is {lowest_performer_value}")
print(f"The average performance score is {average_performance_score}")
print(f"The bonus list of all employees is : {bonus_list}")
print(f"The total bonus of employees are : {total_bonus}")
                

```

```python
class VirtualMachines:

    def __init__(self, name, region, sku, monthly_cost):
        self.name = name
        self.region = region
        self.sku = sku
        self.monthly_cost = monthly_cost

    def calculate_discount(self):
        discount = int(input("Enter the discount you are looking for : "))
        discount_amount = (discount/100) * self.monthly_cost
        print(f"The discount amount is : {discount_amount}")
        return discount_amount

    def final_cost(self):
        final_amount = self.monthly_cost - self.calculate_discount()
        print(f"The final cose of VM after discount is : {final_amount}")

    def display(self):
        print(f"VM name is : {self.name}")
        print(f"VM region is : {self.region}")
        print(f"VM SKU family is : {self.sku}")
    
VM = VirtualMachines("VM01", "East US", "ESv5", 5000)
VM.display()
VM.final_cost()
```

```python
class BankAccount:

    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.balance = balance
    def deposit(self):
        try:
            depositted_amount = int(input("Enter the amount you want to deposit"))
            if depositted_amount > 0:
                self.balance = self.balance + depositted_amount
                print(f"The depositted balance in bank account is {depositted_amount}")
                print(f"The updated balance in bank account is {self.balance}")
            else:
                raise ValueError
        except ValueError:
            print("The amount is invalid to be depositted, enter another amount")
        return depositted_amount
    def withdraw(self):
        try:
            withdrawl_amount = int(input("Enter the amount you want to withdraw"))
            if withdrawl_amount > 0 and withdrawl_amount <= self.balance:
                self.balance = self.balance - withdrawl_amount
                print(f"The withdrawl amount from bank account is {withdrawl_amount}")
                print(f"The updated balance in bank account is {self.balance}")
            elif withdrawl_amount <= 0:
                raise ValueError
            elif withdrawl_amount > self.balance :
                raise ValueError
        except ValueError:
            print("The amount is invalid, enter another amount")
        return withdrawl_amount
    def display(self):
        print(f"Account holder name is : {self.account_holder}")
        print(f"Balance in account is : {self.balance}")

account_holder = BankAccount("Prakhar", 84032)
account_holder.display()
account_holder.deposit()
account_holder.withdraw()
```

```python
class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
    def calaculate_tax(self):
        if self.salary < 50000:
            tax = 0
        elif self.salary >= 50000 and self.salary < 69999:
            tax = 0.1 * self.salary
        elif self.salary >= 70000 and self.salary < 99999:
            tax = 0.2 * self.salary
        else:
            tax = 0.3 * self.salary
        return tax
    def calculate_final_salary(self):
        final_salary = self.salary - self.calaculate_tax()
        print(f"The final salary of employee is : {final_salary}")
    def display(self):
        print(f"Employee name is : {self.name}")
        print(f"basic salary is : {self.salary}")
        print(f"Tax on salary is : {self.calaculate_tax()}")

employee1 = Employee("Prakhar", 84032)
employee1.display()
employee1.calculate_final_salary()
```

```python
class Employee:

    def __init__(self, name, salary, bonus):
        self.name = name
        self.salary = salary
        self.bonus = bonus
    def display(self):
        print(f"Employee name is : {self.name}")
        print(f"basic salary is : {self.salary}")
        print(f"Bonus is : {self.bonus}")
    def calculate_gross(self):
        gross_salary = self.salary + self.bonus
        print(f"Gross salary is : {gross_salary}")


employee1 = Employee("Prakhar", 50000, 5000)
employee1.display()
employee1.calculate_gross()
```
