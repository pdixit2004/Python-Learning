```python
class BankAccount:

    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.balance = balance
    def deposit(self):
        try:
            depositted_amount = int(input("Enter the amount you want to deposit"))
            if depositted_amount > 0:
                self.balance = self.balance + depositted_amount
                print(f"The depositted balance in bank account is {depositted_amount}")
                print(f"The updated balance in bank account is {self.balance}")
            else:
                raise ValueError
        except ValueError:
            print("The amount is invalid to be depositted, enter another amount")
        return depositted_amount
    def withdraw(self):
        try:
            withdrawl_amount = int(input("Enter the amount you want to withdraw"))
            if withdrawl_amount > 0 and withdrawl_amount <= self.balance:
                self.balance = self.balance - withdrawl_amount
                print(f"The withdrawl amount from bank account is {withdrawl_amount}")
                print(f"The updated balance in bank account is {self.balance}")
            elif withdrawl_amount <= 0:
                raise ValueError
            elif withdrawl_amount > self.balance :
                raise ValueError
        except ValueError:
            print("The amount is invalid, enter another amount")
        return withdrawl_amount
    def display(self):
        print(f"Account holder name is : {self.account_holder}")
        print(f"Balance in account is : {self.balance}")

account_holder = BankAccount("Prakhar", 84032)
account_holder.display()
account_holder.deposit()
account_holder.withdraw()
```

```python
class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
    def calaculate_tax(self):
        if self.salary < 50000:
            tax = 0
        elif self.salary >= 50000 and self.salary < 69999:
            tax = 0.1 * self.salary
        elif self.salary >= 70000 and self.salary < 99999:
            tax = 0.2 * self.salary
        else:
            tax = 0.3 * self.salary
        return tax
    def calculate_final_salary(self):
        final_salary = self.salary - self.calaculate_tax()
        print(f"The final salary of employee is : {final_salary}")
    def display(self):
        print(f"Employee name is : {self.name}")
        print(f"basic salary is : {self.salary}")
        print(f"Tax on salary is : {self.calaculate_tax()}")

employee1 = Employee("Prakhar", 84032)
employee1.display()
employee1.calculate_final_salary()
```

```python
class Employee:

    def __init__(self, name, salary, bonus):
        self.name = name
        self.salary = salary
        self.bonus = bonus
    def display(self):
        print(f"Employee name is : {self.name}")
        print(f"basic salary is : {self.salary}")
        print(f"Bonus is : {self.bonus}")
    def calculate_gross(self):
        gross_salary = self.salary + self.bonus
        print(f"Gross salary is : {gross_salary}")


employee1 = Employee("Prakhar", 50000, 5000)
employee1.display()
employee1.calculate_gross()
```
