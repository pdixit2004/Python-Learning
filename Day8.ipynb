```python
numbers = [1,2,3,4,5]

result = []

for n in numbers:
    if n % 2 != 0:
        result.append(n * 2)

print(result)
```

```python
names = ["Raj", "Prakhar", "Amit", "Christopher"]
l_name = ""
for name in names :
    if len(name) > len(l_name) :
        l_name = name
print("The longest name is : " , l_name)
```

```python
numbers = [10,20,10,30,20,40,10]
new_list = []
for num in numbers:
    if num not in new_list:
        new_list.append(num)
print(new_list)
```

```python
numbers = [10,250,8,40,30,100,80,150]
l_num = 0
second_largest = 0
for num in numbers:
    if num > l_num :
        l_num = num
    else:
        second_largest = num
print("Largest num is : " , l_num)
print("Second largest num is : " , second_largest)
```

```python
total = 0
count = 0
numbers = []
for num in range(1,6):
    number = int(input("Enter the numbers in the list : "))
    numbers.append(number)
    total = total + number
print("Total is : " ,total)
average = total / 5
print("Average is : " ,average)
for num in numbers:
    if num > average:
        count += 1
print("Count of numbers greater than average is : " ,count)
```
