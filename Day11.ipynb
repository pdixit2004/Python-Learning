```python
# reverse a list without reverse function
numbers = [10, 20, 30, 40, 50]
stringreversed = []
for i in range(len(numbers)-1,-1,-1):
    stringreversed.append(numbers[i])
print(stringreversed)
```

```python
# count number greater than the average
numbers = [10, 20, 30, 40, 50, 60]
count = 0
sum = 0
bnum = []
for num in numbers :
    sum = sum + num
average = sum / len(numbers)
print("The average of the numbers is : ", average)
for num in numbers:
    if num > average :
        count = count + 1
        bnum.append(num)
print("The number of elements greater than the average is : ", count)
print("The elements greater than the average are : ", bnum)
```

```python
# sum only positive numbers
numbers = [10, -5, 20, -8, 15, -2]
sum = 0
for num in numbers:
    if num > 0:
        sum = sum + num
print("The sum of positive numbers is:", sum)
```

```python
# find even numbers in a list
numbers = [12, 5, 8, 17, 20, 33, 42]
count = 0
for num in numbers :
    if num % 2 == 0 :
        count = count + 1
print("The number of even numbers in the list is : ", count)
```

```python
# find the largest number 
numbers = [10, 250, 7, 90, 42]
lnum = 0
for num in numbers :
    if num  > lnum :
        lnum = num
print("The largest number is : ", lnum)
```
