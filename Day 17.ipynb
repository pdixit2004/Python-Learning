```python
# count the content of file and does not stop when encountered a string
with open("numbers.txt", "w") as file:
    file.write("10\n")
    file.write("25\n")
    file.write("hello\n")
    file.write("15\n")
    file.write("34\n")

def analyze_file(filename):
    countvalidnumbers = 0
    a = 0
    try : 
        with open(filename, "r") as file:
            numbers = file.readlines()
            for num in numbers:
                try:
                    a = int(num)
                    countvalidnumbers = countvalidnumbers + 1
                except ValueError:
                    print("The content is a string")
        return countvalidnumbers
    except FileNotFoundError:
        print("File Not Found")
        return None

result = analyze_file("numbers.txt")
countvalidnumbers = result
print(f"Total count of valid number is : {countvalidnumbers}")


```

```python
# numbers analyzer
with open("numbers.txt", "w") as file:
    file.write("10\n")
    file.write("25\n")
    file.write("40\n")
    file.write("15\n")
    file.write("60\n")
    file.write("5\n")
    file.write("500\n")

def analyze_file(filename):
    lnum = None
    snum = None
    evencount = 0
    oddcount = 0
    total = 0
    try : 
        with open(filename, "r") as file:
            numbers = file.readlines()
            for num in numbers:
                a = int(num)
                total = total + a
                if lnum is None or a > lnum:
                    lnum = a
                if snum is None or snum > a:
                    snum = a
                if a % 2 == 0:
                    evencount = evencount + 1
                if a% 2 != 0:
                    oddcount = oddcount + 1
            average = round(total / len(numbers) , 2)
        return total, average, lnum, snum, evencount, oddcount
    except FileNotFoundError:
        print("File Not Found")
        return None

result = analyze_file("numbers.txt")
total, average, lnum, snum, evencount, oddcount = result

print(f"Total is : {total}")
print(f"Average is : {average}")
print(f"largest num is : {lnum}")
print(f"Smallest num is : {snum}")
print(f"Even count is : {evencount}")
print(f"Odd count is : {oddcount}")


```

```python
# search employees in the file
with open("employees.txt", "r") as file:
    content = file.readlines()
    name = input("Enter the employee you want to search : ")
    found = False
    for line in content:
        if name in line:
            found = True
            break
    if found == True :
            print("Employee Found")
    else:
        print("Employee not found")
```
