```python
def total_valid_transaction(transactions):
    count = 0
    total = 0
    for row in transactions:
        try:
            num = int(row[1])
            #total = total + num
            if num <= 0 :
                raise ValueError("Transactions is invalid. Transaction has to be greater than 0")
            count = count +1
            total = total + num
        except ValueError as errors:
            print(errors)
    return count, total
                

transactions = [
    ["Rahul", 5000],
    ["Priya", 3000],
    ["Amit", -1000],
    ["Neha", 7000],
    ["Ravi", "5000"]
]

total_valid_transaction(transactions)
```
